# Taller 6 - Punto 4: Árboles de decisión

Aplicación: clasificación de vinos a partir de propiedades físico-químicas.

Un árbol de decisión divide los datos mediante reglas jerárquicas. Cada nodo selecciona una característica y un umbral para aumentar la pureza de los grupos resultantes.

```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay

SEED = 42
wine = load_wine(as_frame=True)
X = wine.data
y = wine.target

print("Features:", X.columns.tolist())
print("Clases:", wine.target_names)
display(X.head())
```

```
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=SEED
)

tree = DecisionTreeClassifier(
    criterion="gini",
    max_depth=4,
    min_samples_leaf=3,
    random_state=SEED
)
tree.fit(X_train, y_train)
pred = tree.predict(X_test)

print("Accuracy prueba:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred, target_names=wine.target_names))
```

```
cm = confusion_matrix(y_test, pred)
ConfusionMatrixDisplay(cm, display_labels=wine.target_names).plot()
plt.title("Matriz de confusión - Árbol de decisión")
plt.show()

plt.figure(figsize=(18,9))
plot_tree(tree, feature_names=X.columns, class_names=wine.target_names,
          filled=True, rounded=True, fontsize=8)
plt.title("Árbol de decisión entrenado")
plt.show()
```

```
importancias = pd.Series(tree.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(8,5))
importancias.head(10).sort_values().plot(kind="barh")
plt.xlabel("Importancia")
plt.title("Características más importantes")
plt.show()

muestra = X.median().to_frame().T
muestra.loc[:, "alcohol"] = 13.2
muestra.loc[:, "flavanoids"] = 2.8
muestra.loc[:, "color_intensity"] = 5.0
clase = tree.predict(muestra)[0]
print("Clase predicha:", wine.target_names[clase])
```
